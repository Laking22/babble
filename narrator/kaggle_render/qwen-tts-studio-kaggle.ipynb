2. **Add Input:** upload the `…-kaggle.zip` you exported from Babble as a new **Dataset** and attach it.

4. When it finishes, open the version's **Output** tab and download the `.m4b` (and the `chapters` folder if it stopped early).

Long books: each run stops starting new chapters after about 11 hours (Kaggle's limit is 12). Attach this run's output and run again to continue.
